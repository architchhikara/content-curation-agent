## 1. Setup & Helper Function
Run this cell after installing dependencies.

```python
import os
import sys
from dotenv import load_dotenv

# Add project root to path so we can import scrapers
project_root = os.path.abspath(".")
if project_root not in sys.path:
    sys.path.append(project_root)

from scrapers import get_scraper_for_url

# Load environment variables from .env
load_dotenv()

def test_scraper(url):
    print(f"Testing URL: {url}")
    scraper = get_scraper_for_url(url)
    if not scraper:
        print("Error: No scraper found for this URL.")
        return
    
    print(f"Selected Scraper: {scraper.__class__.__name__}")
    try:
        data = scraper.scrape(url)
        if data:
            print("--- Scraping Results ---")
            for key, value in data.items():
                if key == "metadata":
                    print(f"{key}: {value}")
                else:
                    print(f"{key}: {str(value)[:200]}...")
        else:
            print("Scraping failed: No data returned.")
    except Exception as e:
        print(f"An error occurred: {e}")

print("Setup complete. Helper function 'test_scraper' is ready.")
```

## 2. YouTube Test

```python
urls = [
    "https://www.youtube.com/watch?v=dQw4w9WgXcQ",  # Standard URL
    "https://youtu.be/dQw4w9WgXcQ",               # Shortened URL
    "https://www.youtube.com/shorts/3P9G5L_rV6Q"   # Shorts URL
]

for u in urls:
    test_scraper(u)
    print("-" * 40)
```
